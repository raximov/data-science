# 🎬 MovieLens Dataset — A Data Story

**Dataset**: ml-latest-small · 100 836 ratings · 9 742 movies · 610 users · 3 683 tags

We analyse the **first 1 000 rows** of each CSV file as required by the task.

---
**Story outline**
1. Which years and genres dominate the movie catalog?
2. How do audiences rate movies — distribution & top performers?
3. Which movies are the most controversial?
4. How active and consistent are users?
5. What tags do people leave, and what do they tell us?

```python
# ── Initialise all four analysis classes ──────────────────────────────────────
from movielens_analysis import Movies, Ratings, Tags, Links

DATA = '../datasets/ml-latest-small/'

movies  = Movies(DATA + 'movies.csv')
ratings = Ratings(DATA + 'ratings.csv')
tags    = Tags(DATA + 'tags.csv')
links   = Links(DATA + 'links.csv')

print(f'Movies rows loaded  : {len(movies.rows)}')
print(f'Ratings rows loaded : {len(ratings.rows)}')
print(f'Tags rows loaded    : {len(tags.rows)}')
print(f'Links rows loaded   : {len(links.rows)}')

%timeit -n 1 -r 1 Movies(DATA + 'movies.csv')
```

---
## Chapter 1 · The Catalog — What Years and Genres Do We Have?

Movie titles embed the release year in parentheses, e.g. `Toy Story (1995)`.
Let's see which years contributed the most movies to our sample.

```python
# ── Distribution of movies by release year (top 10) ──────────────────────────
release = movies.dist_by_release()

print('Top 10 years by movie count:')
for year, count in list(release.items())[:10]:
    bar = '█' * count
    print(f'  {year}: {bar} ({count})')

%timeit -n 1 -r 1 movies.dist_by_release()
```

```python
# ── Distribution of movies by genre (top 10) ─────────────────────────────────
genres = movies.dist_by_genres()

print('Top 10 genres by movie count:')
for genre, count in list(genres.items())[:10]:
    bar = '█' * (count // 5)
    print(f'  {genre:<15}: {bar} ({count})')

%timeit -n 1 -r 1 movies.dist_by_genres()
```

```python
# ── Movies with the most genres (top 10) ─────────────────────────────────────
multi = movies.most_genres(10)

print('Top 10 movies with the most genre tags:')
for title, n_genres in multi.items():
    print(f'  {n_genres} genres  →  {title}')

%timeit -n 1 -r 1 movies.most_genres(10)
```

---
## Chapter 2 · The Ratings — How Does the Audience Judge?

Ratings range from 0.5 to 5 stars in 0.5-star increments.
First let's see how ratings are distributed over time and across the rating scale.

```python
# ── Ratings by year ───────────────────────────────────────────────────────────
by_year = ratings.movies.dist_by_year()

print('Ratings per year (first 1 000 ratings):')
for year, count in by_year.items():
    bar = '█' * (count // 2 + 1)
    print(f'  {year}: {bar} ({count})')

%timeit -n 1 -r 1 ratings.movies.dist_by_year()
```

```python
# ── Rating score distribution ─────────────────────────────────────────────────
by_rating = ratings.movies.dist_by_rating()

print('Score distribution (★ = 5 ratings):')
for score, count in by_rating.items():
    bar = '★' * (count // 5 + 1)
    print(f'  {score:.1f}: {bar} ({count})')

%timeit -n 1 -r 1 ratings.movies.dist_by_rating()
```

```python
# ── Top 10 most rated movies (by number of ratings in our sample) ──────────────
top_num = ratings.movies.top_by_num_of_ratings(10)

print('Top 10 most-rated movies:')
for i, (title, cnt) in enumerate(top_num.items(), 1):
    print(f'  {i:>2}. ({cnt} ratings)  {title}')

%timeit -n 1 -r 1 ratings.movies.top_by_num_of_ratings(10)
```

```python
# ── Top 10 highest rated movies (average) ─────────────────────────────────────
top_avg = ratings.movies.top_by_ratings(10, metric='average')

print('Top 10 highest-rated movies (average):')
for i, (title, score) in enumerate(top_avg.items(), 1):
    print(f'  {i:>2}. {score:.2f}  {title}')

%timeit -n 1 -r 1 ratings.movies.top_by_ratings(10, metric='average')
```

```python
# ── Top 10 highest rated movies (median) ──────────────────────────────────────
top_med = ratings.movies.top_by_ratings(10, metric='median')

print('Top 10 highest-rated movies (median):')
for i, (title, score) in enumerate(top_med.items(), 1):
    print(f'  {i:>2}. {score:.2f}  {title}')

%timeit -n 1 -r 1 ratings.movies.top_by_ratings(10, metric='median')
```

---
## Chapter 3 · Controversy — Which Movies Divide Opinion Most?

Variance in ratings reveals how divisive a movie is among the audience.
High variance means some love it while others hate it — classic 'marmite' films.

```python
# ── Top 10 most controversial movies (by rating variance) ─────────────────────
controversial = ratings.movies.top_controversial(10)

print('Top 10 most controversial movies (rating variance):')
for i, (title, var) in enumerate(controversial.items(), 1):
    print(f'  {i:>2}. variance={var:.2f}  {title}')

%timeit -n 1 -r 1 ratings.movies.top_controversial(10)
```

---
## Chapter 4 · Users — Activity and Taste Patterns

Who rates the most? Are users consistent, or do their tastes vary wildly?

```python
# ── How many ratings did each user leave? ─────────────────────────────────────
user_activity = ratings.users.dist_by_ratings_number()

print('Distribution — how many users gave N ratings (first 10 buckets):')
for n_ratings, n_users in list(user_activity.items())[:10]:
    print(f'  {n_ratings} ratings  →  {n_users} user(s)')

%timeit -n 1 -r 1 ratings.users.dist_by_ratings_number()
```

```python
# ── Distribution of users by their average rating (metric='average') ───────────
user_avg_dist = ratings.users.dist_by_ratings_metric(metric='average')

print('Distribution — users by their average rating:')
for avg, n_users in user_avg_dist.items():
    print(f'  avg={avg:.2f}  →  {n_users} user(s)')

%timeit -n 1 -r 1 ratings.users.dist_by_ratings_metric(metric='average')
```

```python
# ── Top 10 most controversial users (high variance in their ratings) ───────────
controversial_users = ratings.users.top_controversial(10)

print('Top 10 most controversial users (by rating variance):')
for i, (uid, var) in enumerate(controversial_users.items(), 1):
    print(f'  {i:>2}. User {uid}  variance={var:.2f}')

%timeit -n 1 -r 1 ratings.users.top_controversial(10)
```

## Chapter 5 · Tags — The Language of Movie Fans

Tags are free-text metadata written by users. They range from one-word labels
like `'funny'` or `'sci-fi'` to elaborate descriptive phrases.
Let's explore what our sample reveals about the community's vocabulary.

```python
# ── Most popular tags (by frequency of use) ───────────────────────────────────
popular = tags.most_popular(10)

print('Top 10 most-used tags:')
for i, (tag, cnt) in enumerate(popular.items(), 1):
    print(f'  {i:>2}. "{tag}"  used {cnt} time(s)')

%timeit -n 1 -r 1 tags.most_popular(10)
```

```python
# ── Tags with the most words ──────────────────────────────────────────────────
wordy = tags.most_words(10)

print('Top 10 unique tags by word count:')
for i, (tag, wc) in enumerate(wordy.items(), 1):
    print(f'  {i:>2}. ({wc} words) "{tag}"')

%timeit -n 1 -r 1 tags.most_words(10)
```

```python
# ── Longest tags by character count ──────────────────────────────────────────
longest = tags.longest(5)

print('Top 5 longest unique tags (characters):')
for i, tag in enumerate(longest, 1):
    print(f'  {i}. ({len(tag)} chars) "{tag}"')

%timeit -n 1 -r 1 tags.longest(5)
```

```python
# ── Tags that are both wordy AND long (intersection) ──────────────────────────
both = tags.most_words_and_longest(10)

print(f'Tags in both top-10 most-words and top-10 longest ({len(both)} found):')
for tag in both:
    print(f'  "{tag}"')

%timeit -n 1 -r 1 tags.most_words_and_longest(10)
```

```python
# ── Tags containing a specific word ──────────────────────────────────────────
sci_tags = tags.tags_with('sci')

print(f'Tags containing "sci" ({len(sci_tags)} found):')
for tag in sci_tags:
    print(f'  "{tag}"')

%timeit -n 1 -r 1 tags.tags_with('sci')
```

## Chapter 6 · Links — Connecting to the Wider World

Each movie in the dataset has identifiers for both IMDB and TMDB.
The `Links` class can use these to enrich our data from external sources.

```python
# ── Fetch IMDB info for the first 3 movies in our sample ──────────────────────
sample_ids = [row['movieId'] for row in links.rows[:3]]
imdb_data = links.get_imdb(sample_ids, ['Director'])

print('IMDB Director data for a sample of movies:')
for row in imdb_data:
    print(f'  movieId={row[0]}  Director: {row[1]}')

%timeit -n 1 -r 1 links.get_imdb(sample_ids, ['Director'])
```
