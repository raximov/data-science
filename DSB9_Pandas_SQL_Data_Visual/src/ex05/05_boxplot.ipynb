```python
import pandas as pd
import sqlite3
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('../data/ab-test.csv')
df
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 8), sharey=True)

bp_styles = {
    'boxprops': {'color': 'darkgreen', 'linewidth': 3},
    'medianprops': {'color': 'red', 'linewidth': 2},
    'whiskerprops': {'color': 'red', 'linewidth': 1},
    'capprops': {'color': 'red', 'linewidth': 1},
    'showmeans': True}

df[df['time']=='after'].boxplot(column='diff', by='group', ax=ax1, **bp_styles)
ax1.set_title('after', fontsize=12)

df[df['time']=='before'].boxplot(column='diff', by='group', ax=ax2, **bp_styles)
ax2.set_title('before', fontsize=12)

plt.suptitle('Lag before labs deadlines. After and before the Newsfeed', fontsize=15)
plt.show()
```
