## 1. Предобработка

1. Создайте тот же DataFrame, что и в предыдущем упражнении.
2. Используйте `train_test_split` с параметрами `test_size=0.2`, `random_state=21`, чтобы получить `X_train`, `y_train`, `X_test`, `y_test`. Используйте дополнительный параметр `stratify`.

````python
# Загрузка данных
df = pd.read_csv('../data/ex01/day-of-week-not-scaled.csv')
df_day = pd.read_csv('../data/ex00/dayofweek.csv')
df['dayofweek'] = df_day['dayofweek']
````

````python
X = df.drop('dayofweek', axis=1)
y = df['dayofweek']
````

````python
# train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=21, stratify=y
)
````

## 2. SVM

1. Используйте лучшие параметры из предыдущего упражнения и обучите модель SVM.
2. Необходимо вычислить `accuracy`, `precision`, `recall`, `ROC AUC`.

 - `precision` и `recall` должны быть рассчитаны для каждого класса (используйте `average='weighted'`)
 - `ROC AUC` должен быть рассчитан для каждого класса против всех остальных (все возможные попарные комбинации), затем взять взвешенное среднее для итоговой метрики
 - код в ячейке должен выводить результат в таком виде:

```

```

````python
# 2. SVM: обучение и метрики
svm = SVC(
    kernel='linear',      # лучшие параметры из ex01
    C=1,
    gamma='scale',
    probability=True,     # обязательно для roc_auc_score
    random_state=21
)
svm.fit(X_train, y_train)
y_pred = svm.predict(X_test)
y_proba = svm.predict_proba(X_test)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='weighted')
recall = recall_score(y_test, y_pred, average='weighted')
roc_auc = roc_auc_score(
    pd.get_dummies(y_test), y_proba, average='weighted', multi_class='ovr'
)

print(f"accuracy is {accuracy:.5f}")
print(f"precision is {precision:.5f}")
print(f"recall is {recall:.5f}")
````

````python
# Предсказания и вероятности для тестовой выборки
y_pred = svm.predict(X_test)
y_proba = svm.predict_proba(X_test)

# Вычисление accuracy, precision, recall
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='weighted')
recall = recall_score(y_test, y_pred, average='weighted')

# Вычисление accuracy, precision, recall
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='weighted')
recall = recall_score(y_test, y_pred, average='weighted')

# Вычисление ROC AUC (one-vs-rest, average='weighted')
roc_auc = roc_auc_score(
    pd.get_dummies(y_test), y_proba, average='weighted', multi_class='ovr'
)
````

````python
# Вывод результатов
print(f"accuracy is {accuracy:.5f}")
print(f"precision is {precision:.5f}")
print(f"recall is {recall:.5f}")
print(f"roc_auc is {roc_auc:.5f}")
````

## 5. Предсказания

2. Проанализируйте: для какого дня недели (`weekday`) ваша модель делает больше всего ошибок (в % от общего числа примеров этого класса в полном датасете), для какого `labname` и для каких `users`.
3. Сохраните модель.

````python
# 1. Выбор лучшей модели по accuracy (если равны — по precision)
metrics = [
    {'name': 'svm', 'accuracy': accuracy, 'precision': precision, 'model': svm},
    {'name': 'tree', 'accuracy': accuracy_tree, 'precision': precision_tree, 'model': tree},
    {'name': 'rf', 'accuracy': accuracy_rf, 'precision': precision_rf, 'model': rf}
]
best = sorted(metrics, key=lambda x: (x['accuracy'], x['precision']), reverse=True)[0]
print(f"Лучшая модель: {best['name']}")
````

````python
# 2.1. Анализ ошибок по дням недели (weekday)
from sklearn.metrics import confusion_matrix

best['model'].fit(X_train, y_train)
y_pred_best = best['model'].predict(X_test)
cm = confusion_matrix(y_test, y_pred_best, labels=np.unique(y_test))
errors = (cm.sum(axis=1) - np.diag(cm)) / cm.sum(axis=1)
for i, day in enumerate(np.unique(y_test)):
    print(f'Доля ошибок для дня недели {day}: {errors[i]:.2%}')
````

````python
# 2.2. Анализ ошибок по labname
lab_cols = [col for col in X_test.columns if col.startswith('labname_')]
for lab in lab_cols:
    idx = X_test[lab] == 1
    if idx.sum() == 0:
        continue
    err = (y_test[idx] != y_pred_best[idx]).sum() / idx.sum()
    print(f'Доля ошибок для {lab}: {err:.2%}')
````

````python
# 2.3. Анализ ошибок по users
user_cols = [col for col in X_test.columns if col.startswith('uid_user_')]
for user in user_cols:
    idx = X_test[user] == 1
    if idx.sum() == 0:
        continue
    err = (y_test[idx] != y_pred_best[idx]).sum() / idx.sum()
    print(f'Доля ошибок для {user}: {err:.2%}')
````

````python
# 3. Сохранение лучшей модели
joblib.dump(best['model'], '../data/ex02/best_model.joblib')
print("Лучшая модель сохранена в '../data/ex02/best_model.joblib'")
````

## 6. Функция

1. Напишите функцию, которая принимает список разных моделей и соответствующий список параметров (dict) и возвращает dict, содержащий все 4 метрики для каждой модели.

````python
# Функция для вычисления метрик для списка моделей и их параметров
def evaluate_models_with_params(models_with_params, X_test, y_test):
    """
    models_with_params: список кортежей (name, model, params_dict)
    X_test, y_test: тестовые данные
    Возвращает словарь с accuracy, precision, recall, roc_auc для каждой модели
    """
    results = {}
    y_test_bin = pd.get_dummies(y_test)
    for name, model, params in models_with_params:
        # Если нужно, можно установить параметры модели:
        if params:
            model.set_params(**params)
        y_pred = model.predict(X_test)
        if hasattr(model, "predict_proba"):
            y_proba = model.predict_proba(X_test)
        else:
            y_proba = None
        accuracy = accuracy_score(y_test, y_pred)
        precision = precision_score(y_test, y_pred, average='weighted')
        recall = recall_score(y_test, y_pred, average='weighted')
        roc_auc = roc_auc_score(
            y_test_bin, y_proba, average='weighted', multi_class='ovr'
        ) if y_proba is not None else None
        results[name] = {
            'accuracy': accuracy,
            'precision': precision,
            'recall': recall,
            'roc_auc': roc_auc,
            'params': params
        }
    return results

# Пример использования:
# models_with_params = [
#     ('svm', svm, {'kernel': 'linear', 'C': 1, 'gamma': 'scale'}),
#     ('tree', tree, {'max_depth': 8, 'criterion': 'entropy'}),
#     ('rf', rf, {'n_estimators': 100, 'max_depth': 8, 'criterion': 'entropy'})
# ]
# metrics_dict = evaluate_models_with_params(models_with_params, X_test, y_test)
# print(metrics_dict)
````

````python
# Пример использования функции
models_with_params = [
    ('svm', svm, {'kernel': 'linear', 'C': 1, 'gamma': 'scale'}),
    ('tree', tree, {'max_depth': 8, 'criterion': 'entropy'}),
    ('rf', rf, {'n_estimators': 100, 'max_depth': 8, 'criterion': 'entropy'})
]

metrics_dict = evaluate_models_with_params(models_with_params, X_test, y_test)
for name, metrics in metrics_dict.items():
    print(f"{name}: {metrics}")
````
