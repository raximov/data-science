```python
import pandas as pd
import sqlite3
```

1. Put the database in the data subfolder in the src directory.

```python
db_file = '../data/checking-logs.sqlite'
```

2. Create a connection to the database using the sqlite3 library.

```python
conn = sqlite3.connect(db_file)
```

3. Get the schema of the pageviews table using pd.io.sql.read_sql() and the query "PRAGMA table_info(pageviews);"

```python
query_schema = "PRAGMA table_info(pageviews);"
schema = pd.read_sql(query_schema, conn)
print("Schema:")
print(schema)
```

4. Get only the first ten rows of the pageviews table to see what the table looks like.

```python
query_first_ten = "SELECT * from pageviews LIMIT 10"
result_first_ten = pd.read_sql(query_first_ten, conn)
print("\nFirst Ten Rows from pageviews Table:")
print(result_first_ten)
```

5. Use one query to get the subtable where:
- Only "uid" and "datetime" are used.
- Only user data (user_*), not admin data, is used.
- It is sorted by "uid" in ascending order.
- The index column is "datetime".
- "datetime" is converted to a DatetimeIndex.
- The name of the dataframe is pageviews.

```python
print("\nCustom Subtable:")
query_custom = "SELECT uid, datetime from pageviews where uid LIKE 'user_%' ORDER BY uid ASC;" 
pageviews = pd.read_sql(query_custom, conn, parse_dates=['datetime'], index_col='datetime')
print(pageviews)
```

6. Close the connection to the database.

```python
conn.close()
```
