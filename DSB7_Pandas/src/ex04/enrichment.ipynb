```python
import pandas as pd
import numpy as np
# import requests --- IGNORE --- because .json files are already in the datasets folder
```

```python
df = pd.read_json("../ex02/auto.json")
```

```python
pd.options.display.float_format = "{:.2f}".format
```

```python
df.isna().sum()
```

```python
sample_cars = df[["CarNumber", "Make", "Model"]].sample(
    n=200, replace=True, random_state=21
)

sample_refund = (
    df["Refund"].sample(n=200, replace=True, random_state=21).reset_index(drop=True)
)

sample_fines = (
    df["Fines"].sample(n=200, replace=True, random_state=21).reset_index(drop=True)
)

sample200 = sample_cars.reset_index(drop=True)
sample200["Refund"] = sample_refund
sample200["Fines"] = sample_fines

sample200
```

```python
concat_rows = pd.concat([df, sample200], axis=0, ignore_index=True)

concat_rows
```

```python
np.random.seed(21)

year_series = pd.Series(
    np.random.randint(1980, 2020, size=len(concat_rows)), name="Year"
)

fines = pd.concat([concat_rows.reset_index(drop=True), year_series], axis=1)

fines
```

```python
surname = pd.read_json("../../datasets/surname.json")
surname.columns = surname.iloc[0]
surname = surname[1:].reset_index(drop=True)
surname
```

```python
random_surnames = (
    surname["NAME"]
    .sample(n=int(len(fines["CarNumber"].unique())), random_state=21, replace=True)
    .reset_index(drop=True)
)
```

```python
owners = pd.DataFrame(
    {"CarNumber": fines["CarNumber"].unique(), "SURNAME": random_surnames}
)

owners
```

```python
new_fines = pd.DataFrame(
    {
        "CarNumber": [
            "MYU0000000",
            "MYU1111111",
            "MYU2222222",
            "MYU3333333",
            "MYU4444444",
        ],
        "Refund": [2, 1, 9, 2, 6],
        "Fines": [12500.00, 12500.00, 18500.00, 12500.00, 12500.00],
        "Make": ["GM", "GM", "GM", "GM", "GM"],
        "Model": ["Cobalt", "Cobalt", "Gentra", "Cobalt", "Cobalt"],
        "Year": [2013, 2023, 2019, 2024, 2025],
    }
)

fines = pd.concat([fines, new_fines], ignore_index=True)
```

```python
# del last 20 row
owners = owners.iloc[:-20] if len(owners) > 20 else owners
```

```python
new_owners = pd.DataFrame(
    {
        "CarNumber": ["X111XX", "Y222YY", "Z333ZZ"],
        "SURNAME": ["TAYLOR", "HARRIS", "MARTIN"],
    }
)

owners = pd.concat([owners, new_owners], ignore_index=True)
```

### Inner join


```python
pd.merge(fines, owners, on="CarNumber", how="inner")
```

### Outer join


```python
pd.merge(fines, owners, on="CarNumber", how="outer")
```

### Left join


```python
pd.merge(fines, owners, on="CarNumber", how="left")
```

### Right join


```python
pd.merge(fines, owners, on="CarNumber", how="right")
```

```python
pivot = pd.pivot_table(
    fines,
    index=["Make", "Model"],
    columns="Year",
    values="Fines",
    aggfunc="sum",
    fill_value=float("nan"),
)

pivot.columns = pivot.columns.astype(str)

pivot
```

```python
fines.to_csv("fines.csv", index=False)


owners.to_csv("owners.csv", index=False)
```
