```python
import pandas as pd
```

```python
views = pd.read_csv("../data/feed-views.log", sep="\t", names=["datetime", "user"])
views
```

```python
views["datetime"] = pd.to_datetime(views["datetime"])
views.dtypes
```

```python
views["year"] = views["datetime"].dt.year
views["month"] = views["datetime"].dt.month
views["day"] = views["datetime"].dt.day
views["hour"] = views["datetime"].dt.hour
views["minute"] = views["datetime"].dt.minute
views["second"] = views["datetime"].dt.second
views
```

```python
bins = [0, 4, 7, 11, 17, 20, 24]
labels = ["night", "early morning", "morning", "afternoon", "early evening", "evening"]
views["daytime"] = pd.cut(views["hour"], bins=bins, labels=labels, right=False)
views
```

```python
views = views.set_index("user")
views
```

```python
print(views[["datetime", "hour", "daytime"]].head())
```

```python
views.count()
```

```python
views["daytime"].value_counts()
```

```python
views["daytime"].value_counts().reindex(
    ["night", "early morning", "morning", "afternoon", "early evening", "evening"]
)
```

```python
views_sorted = views.sort_values(by=["hour", "minute", "second"], ascending=True)
views
```

```python
min_hour = views["hour"].min()
max_hour = views["hour"].max()
print(f"\nMin hour: {min_hour}, Max hour: {max_hour}")
```

```python
mode_daytime = views["daytime"].mode()[0]
print("Mode daytime:", mode_daytime)
```

```python
max_night_hour = views.loc[views["daytime"] == "night", "hour"].max()
print("Max night hour:", max_night_hour)
```

```python
min_morning_hour = views.loc[views["daytime"] == "morning", "hour"].min()
print("Min morning hour:", min_morning_hour)
```

```python
# Night user
night_user_example = views.loc[views["hour"] == max_night_hour].head(1)
night_user_example
```

```python
# Morning user
morning_user_example = views.loc[views["hour"] == min_morning_hour].head(1)
morning_user_example
```

```python
# Three earliest hours
earliest = views.nsmallest(3, "hour")[["hour", "daytime"]]
earliest
```

```python
# Three latest hours
latest = views.nlargest(3, "hour")[["hour", "daytime"]]
latest
```

```python
stats = views.describe()
stats
```

```python
q1 = stats.loc["25%", "hour"]
print("\nQ1:", q1)

q3 = stats.loc["75%", "hour"]
print("Q3:", q3)

iqr = q3 - q1
print("IQR(hours):", iqr)
```
